# Practical-4 : Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets.


Step 1 : Import libraries

```python
# Import library.
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2 : Create Dataset

```python
# Create a dictionary containing student details such as Month , Units_sold , Revenue, Profit.
data={
    "Month" : ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],
    "Units_Sold" : [120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue" : [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Profit" : [20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
```

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)

df
```

Step 4 : Generate Descriptive Statistics

```python
# Generative describative statistic for all numerical column.
df.describe()
```

Step 5 : Calculate Mean

```python
# Calculate the average number of unit sold per month.
average_units = df['Units_Sold'].mean()

# display the result.
print("Average Units Sold:", average_units)
```

Step 6 : Calculate the middle value

```python
# Calculate the middle value of the Units_sold column.
median_units = df['Units_Sold'].median()

# display the result.
print("Median Units Sold:", median_units)
```

Step 7 : Calculate Maximum and Minimum

```python
# Find the highest number of units sold in a month.
maximum_units = df['Units_Sold'].max()

# Find the slowest number of units sold in a month.
minimum_units = df['Units_Sold'].min()

# display the result.
print("Maximum Units Sold:", maximum_units)
print("Minimum Units Sold:", minimum_units)
```

Step 8 : Calculate Standard Deviation

```python
# Calculate the standard deviation of units sold.
std_units = df['Units_Sold'].std()

# display the result.
print("Standard Deviation of Units Sold:", std_units)
```

Step 9 : find the Month with Highest sales

```python
# Find the index of the row having the maximum revenue.
highest_sales_index = df['Revenue'].idxmax()

# Retrieve the month corresponding to that index.
highest_sales_month = df.loc[highest_sales_index, 'Month']

# Retrive the maximum revenue.
highest_revenue = df.loc[highest_sales_index, 'Revenue']

# display the result.
print("Month with Highest Sales:", highest_sales_month)
print("Highest Revenue:", highest_revenue)
```

Step 10 : find the Month with lowest sales

```python
# Find the index of the row having the minimum revenue.
lowest_sales_index = df['Revenue'].idxmin()

# Retrieve the month corresponding to that index.
lowest_sales_month = df.loc[lowest_sales_index, 'Month']

# Retrieve the minimum revenue.
lowest_revenue = df.loc[lowest_sales_index, 'Revenue']

# display the result.
print("Month with Lowest Sales:", lowest_sales_month)
print("Lowest Revenue:", lowest_revenue)
```

Step 11 : Calculate Total Revenue

```python
# Calculate the total revenue generated during the year.
total_revenue = df['Revenue'].sum()

# display the result.
print("Total Revenue:", total_revenue)
```

Step 12 : Calculate Total Profit

```python
# Calculate the total profit generated during year.
total_profit = df['Profit'].sum()

# display the result.
print("Total Profit:", total_profit)
```

Step 13 : Identify The Sales Trend

```python
from matplotlib.typing import LineStyleType
# Create  a line chart to visualize monthly revenue.
plt.figure(figsize=(10,5))

# plot Month on the X-axis and Revenue on the Y-axis.
plt.plot(df["Month"], df["Revenue"], marker="o",  linestyle=":" )

# Add a title to the graph.
plt.title("Monthly Revenue")

# Add labels to the X-axis and Y-axis.
plt.xlabel("Month")
plt.ylabel("Revenue")

# Rotate xticks 45.
plt.xticks(rotation=45)

# Display the graph.
plt.show()
```

Srep 14 : Analyze Profit Trend

```python
# Create a line chart for monthly profit.
plt.figure(figsize=(10,5))

# plot Month on the X-axis and Profit on the Y-axis.
plt.plot(df["Month"], df["Profit"], marker="o", )

# Add a title to the graph.
plt.title("Monthly Profit")

# Add labels to the X-axis and Y-axis.
plt.xlabel("Month")
plt.ylabel("Profit")

# Rotate xticks 45.
plt.xticks(rotation=45)

# Display the graph.
plt.show()

```

Step 15: Compare Revanue And Profit

```python
# Create a line chart to compare Revenue and Profit.
plt.figure(figsize=(10,5))

# Plot Revenue.
plt.plot(df["Month"], df["Revenue"], marker="o", label="Revenue")

# Plot Profit.
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

# Add labels to the X-axis and Y-axis.
plt.xlabel("Month")
plt.ylabel("Amount")

# Add a title to the graph.
plt.title("Revenue vs Profit")

plt.legend()

# Rotate xticks 45.
plt.xticks(rotation=45)

# Display the graph.
plt.show()
```

Step 16 : Identify Growth in Sales

```python
# Calculate the percentage change in revenue.
df["Revenue_Growth_%"] = df["Revenue"].pct_change() * 100

# Display the updated dataset.
df
```
