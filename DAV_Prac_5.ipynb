### Practical 5: Analyze datasets using correlation, covariance, outlier detection, and distribution analysis.

Step 1 : Import required library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Step 2: Create a dataset

```python
# Create a dictionary containing student details such as Student_ID,Study_Hours, Attendence, Assignment-score, Final_Marks.
data = {
    "Student_ID" : [101,102,103,104,105,106,107,108,109,110,111,112,113,114,115],
    "Study_Hours" : [2,3,4,5,6,7,8,9,3,4,5,6,7,10,1],
    "Attendence" : [65,70,75,80,82,85,90,92,68,74,78,84,88,95,40],
    "Assignment-score" : [55,60,65,70,72,78,82,88,62,67,71,76,80,95,35],
    "Final_Marks" : [50,55,62,68,72,78,82,90,58,63,69,75,81,98,25]
}
```

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)
df
```

Step 3 : Inspect the dataset

```python
# Display the first 5 record of dataset
df.head()
```

```python
# display infromation about dataset.
df.info()
```

```python
# display the number of rows and column.
df.shape
```

```python
# Generative describative statistic for all numerical column.
df.describe()
```

Step 4: Calculate Co-relation

```python
# Calculate the co-relation between study hours and final marks.
correlation = df['Study_Hours'].corr(df['Final_Marks'])

print("Correlation coefficient:", correlation)
```

Step 5: Calculate Co-relation Matrix

```python
# Calculate co-relation among by numerical amount.
correlation_matrix = df.corr()
print(correlation_matrix)
```

```python
# Display the correlation matrix.
correlation_matrix
```

Step 6: Visualize Correlation Using Scatter Plot

```python
# Create a scatter plot to visualize the relationship between Study_Hours and Final_Marks.
plt.scatter(df['Study_Hours'], df['Final_Marks'])
plt.xlabel('Study_Hours')
plt.ylabel('Final_Marks')
plt.title('Scatter Plot of Study_Hours vs Final_Marks')
plt.show()
```

Step 7: Calculate Covarience

```python
# Calculate covarience bewtween Study Hours and Final Marks.
covarience = np.cov(df['Study_Hours'], df['Final_Marks'])
print("Covarience:", covarience)
```

Step 8: Calculate Covarience Matrix

```python
# Select the numerical columns for covarience analysis.
numerical_columns = df.select_dtypes(include=[np.number])
# Calculate the covarience matrix for the selected columns.
covarience_matrix = numerical_columns.cov()
```

```python
# Calculate the covarience matrix.
covarience_matrix
```

Step 9: Detect Outliers Using IQR(Select the Final_Marks column for outlier analysis)

```python
Q1 = df['Final_Marks'].quantile(0.25)
Q3 = df['Final_Marks'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

Step 10: Identify Outliers

```python
# Identify Outliers.
outliers = df[
    (df['Final_Marks'] < lower_bound) |
    (df['Final_Marks'] > upper_bound)
]

print("Outliers:")
print(outliers)
```

Step 11: Count Outliers(Count the total number of detected outliers.)

```python
outlier_count = len(outliers)
print("Total number of outliers:", outlier_count)
```

Step 12: Visualize outliers using box plot

```python
plt.boxplot(df['Final_Marks'])
plt.ylabel('Final_Marks')
plt.title('Box Plot of Final_Marks')
plt.show()
```

Step 13: create a histogram (Create a histogram to observe the distribution of final marks)

```python
# Create a histogram to observe the distribution of final marks.
plt.hist(df["Final_Marks"], bins=10, edgecolor="black")
plt.xlabel("Final_Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Final_Marks")
plt.show()
```

Step 14: Analyze distribution using skewness

```python
skewness = df['Final_Marks'].skew()
print("Skewness:", skewness)
```

Step 15: Calculate Kurtosis

```python
kurtosis = df['Final_Marks'].kurtosis()
print("Kurtosis:", kurtosis)
```

Step 16: Distribution of study hours(create a hsitogram for study hours.)

```python
# create a hsitogram for study hours.
plt.hist(df["Study_Hours"], bins=10, edgecolor="black")
plt.xlabel("Study_Hours")
plt.ylabel("Frequency")
plt.title("Histogram of Study_Hours")
plt.show()
```

Step 17: compare Variable Using Box Plots (Create box plots for multiple numerical variables.)

```python
# Select numerical variables
numerical_data = df[
            ["Study_Hours", "Attendence","Assignment-score", "Final_Marks"]
]
# Create box plots
plt.figure(figsize=(10, 6))
plt.boxplot(
numerical_data.values,
labels=numerical_data.columns
)
plt.title("Comparison of Numerical Variables")
plt.xlabel("Variables")
plt.ylabel("Values")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
