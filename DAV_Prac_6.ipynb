### Practical-6 : Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).

Step 1 : Import Libraries

```python
# Import library.
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2 : create a dataset

```python
# Create a dictionary containing student details such as Month , Units_sold , product, Region, sales.
data={
    'Month' : ['Jan', 'Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Product': ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    'Region': ['North','South','East','West','North','South','East','West','North','South','East','West'],
    'Sales': [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Units_Sold' : [12,30,25,14,35,30,18,38,32,21,45,40]
}
```

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)
df
```

Step 3 : Create a Line chart

```python
# How did the company's sales change from january to December?
plt.figure(figsize=(10,6))
plt.plot(df['Month'], df['Sales'])
# Add title and labels
plt.title('Company Sales Over Time')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

# Interpretations :
The line chart shows the company’s sales from January to December.
Sales generally increase throughout the year, starting at ₹1,20,000 in January and reaching the highest value of ₹2,50,000 in December.

Step 4 : Create a Bar chart

```python
# which product generated how much total sales?
plt.figure(figsize=(10,6))
plt.bar(df['Product'], df['Sales'])
# Add title and labels
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

# Interpretations :
The bar chart compares sales values for different products.
The dataset contains Laptop, Mobile, and Tablet products.
Tablet is the best-performing product, while Laptop has the lowest total sales among the three.

Step 5 : Create a Histogram

```python
# How are the individual sales values distributed?
plt.figure(figsize=(10,6))
plt.hist(df['Sales'], bins=10, color='skyblue', edgecolor='black')
# Add title and labels
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

# Interpretations :
he histogram shows how the individual sales values are distributed.
The sales values range from ₹1,20,000 to ₹2,50,000

Step 6 : Create a Scatter Plot

```python
# Is there a relationship between the number of units sold and total sales?
# Create scatter plot
plt.figure(figsize=(10,6))
plt.scatter(df['Units_Sold'], df['Sales'])
# Add title and labels
plt.title('Relationship Between Units Sold and Total Sales')
plt.xlabel('Number of Units Sold')
plt.ylabel('Total Sales')
# Add grid
plt.grid(True)
plt.show()
```

# Interpretations :
The scatter plot shows the relationship between the number of units sold and total sales.
The points generally move upward from left to right.

Step 7 : Create a Pie Chart

```python
# What percentage of total sales comes from each region?
# Create pie chart
plt.figure(figsize=(10,6))
region_sales = df.groupby('Region')['Sales'].sum()
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', startangle=90)
# Add title
plt.title('Percentage of Total Sales by Region')
plt.axis('equal')
plt.show()
```

# Interpretations :
The pie chart shows the contribution of each region to the company’s total sales.
West contributes the highest percentage (26.71%), followed very closely by East (26.48%).
South contributes 24.59%, while North has the lowest contribution at 22.22%.
