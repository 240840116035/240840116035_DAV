Practical-8 : Create interactive visualizations using Plotly Express and Plotly Graph Objects.


Step 1 : Import required libraries

```python
# Import Pandas for data manipulation
import pandas as pd

# Import Plotly Express for quick and easy visualizations
import plotly.express as px

# Import Plotly Graph Objects for customized visualizations
import plotly.graph_objects as go
```

Step 2 : Create the common dataset

```python
# Create a dictionary containing student details such as Month , Units_sold , product, Region, sales.
data={
    'Month' : ['Jan', 'Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Product': ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    'Region': ['North','South','East','West','North','South','East','West','North','South','East','West'],
    'Sales': [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Units_Sold' : [12,30,25,14,35,30,18,38,32,21,45,40],
    'Profit' : [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}
```

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)
df
```

```python
# Display the first five records.
df.head()
```

```python
# Display the number of rows and columns.
df.shape
```

```python
# Display columns names and data types.
df.info()
```

Step 3 : Create an Interactive Line chart Using Plotly Express

```python
# How does sales revenue change throught the year?
figure = px.line(df, x="Month", y="Sales", title="Sales Revenue Through the Year")
figure.show()
```

Step 4 : Improve the Hover Information.

```python
# Create a line chart using Sales over Month.
# Add additional columns to the bevor information. Students can see Product, Region, Region, Units Sold abd Profit when they move the mouse over a data point.
figure = px.line(df, x="Month", y="Sales", title="Sales Revenue Through the Year", color='Product', markers=True)
figure.show()
```

Step 5 : Create an Interactive Bar Chart

```python
# Group the dataset according to product.
figure = px.bar(df, x="Month", y="Units_Sold", color="Product", title="Units Sold by Product")
figure.show()
```

```python
# Calculate the total sales generated by each product.
figure = px.bar(df, x="Product", y="Sales", color="Product", title="Total Sales by Product")
figure.show()
```

```python
# Create an interactive bar chart.
figure = px.bar(df, x="Product", y="Sales", color="Region", title="Total Sales by Product and Region")
figure.show()
```

Step 6 : Interactive Scatter Plot

```python
# To analyze the relationship between Units Sold and sales using an interactive Scatter Plot.
figure = px.scatter(df, x="Units_Sold", y="Sales", color="Product", size="Profit", title="Relationship between Units Sold and Sales")
figure.show()
```

Step 7 : Interactive Scatter Plot with Color

```python
# We will use Region as the color variable.
figure = px.scatter(df, x="Units_Sold", y="Sales", color="Region", size="Profit", title="Relationship between Units Sold and Sales")
figure.show()
```

Step 8 : Create an Interactive Pie chart

```python
# To visualize regional sales contribution using an interactive Pie chart.
figure = px.pie(df, values="Sales", names="Region", hole=0, title="Regional Sales Contribution")
figure.show()
```

Step 9: Plotly Graph Objects

```python
# To create an interactive visualization using Plotly Graph Objects and understand how it differs from Plotly Express.
figure = go.Figure(data=[go.Scatter(x=df.Month, y=df.Sales, mode='lines+markers',)])
figure.show()
```

Step 10 : Create Line Chart using graph objects

```python
# Create a Figure object.
figure = go.Figure()
```

```python
# Add a line trace to the figure. This trace represents the monthly sales data.
figure.add_trace(go.Scatter(x=df.Month, y=df.Sales, mode='lines+markers', name='Sales'))
```
