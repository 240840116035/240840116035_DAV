Step 5 : Check missing values and duplicates

```python
print("Missing values:")
print(df.isnull().sum())

print("Duplicate rows:")
print(df.duplicated().sum())
```

Step 6 : Handle missing values and remove duplicates

```python
# Handle missing values
for col in df.columns:

    if df[col].dtype == "object":
        df[col] = df[col].fillna("Unknown")

    else:
        df[col] = df[col].fillna(df[col].median())

# Remove duplicate records
df = df.drop_duplicates()

print("Missing values after:", df.isnull().sum().sum())
print("Duplicates after:", df.duplicated().sum())
print("Shape after:", df.shape)
```

Step 7 : Data transformation

```python
# Rename columns
df = df.rename(columns={
    "Financial Loss (Million $)": "Financial_Loss_Million",
    "Number of Affected Users": "Affected_Users",
    "Attack Type": "Attack_Type",
    "Target Industry": "Target_Industry"
})

# Fix data types
df["Year"] = df["Year"].astype(int)

df["Financial_Loss_Million"] = df[
    "Financial_Loss_Million"
].astype(float)

df["Affected_Users"] = df[
    "Affected_Users"
].astype(int)

# Sort dataset
df = df.sort_values(
    by=["Year", "Country"]
).reset_index(drop=True)

df.head()
```

Step 8 : Export cleaned dataset

```python
df.to_csv("cybersecurity_threats_cleaned.csv", index=False)

from google.colab import files

files.download("cybersecurity_threats_cleaned.csv")
```
